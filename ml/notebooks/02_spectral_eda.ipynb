# TerraSpectra - Spectral Exploratory Data Analysis

This notebook reviews per-band statistics generated by the windowed spectral EDA script. Run it only after inspecting an accessible real scene.

The statistics are descriptive and do not establish crop disease labels.

```python
from pathlib import Path
import csv
import json

summary_path = Path('../../docs/dataset/spectral-eda/spectral-summary.json')
csv_path = Path('../../docs/dataset/spectral-eda/spectral-band-statistics.csv')

if not summary_path.exists():
    raise FileNotFoundError('Run ml/scripts/spectral_eda.py with a real scene first.')

summary = json.loads(summary_path.read_text(encoding='utf-8'))
print('Scene:', summary['scene'])
print('Dimensions:', summary['width'], 'x', summary['height'])
print('Bands:', summary['band_count'])
```

```python
with csv_path.open(newline='', encoding='utf-8') as file:
    band_statistics = list(csv.DictReader(file))

for band in band_statistics:
    print(band)

print('Constant bands:', [b['band'] for b in band_statistics if b['constant'] == 'True'])
```

## Interpretation checklist

- Check missing or non-finite observations.
- Review constant bands and possible noisy bands.
- Confirm wavelength mapping from original metadata.
- Do not infer disease classes from spectral statistics alone.
- Fit normalization and PCA on training data only.
